```python
import numpy as np
import pandas as pd
!pip install matplotlib
!pip install tensorflow
import tensorflow as tf
 # defining Train and Test data path
train_path = r"C:\Users\user\Downloads\DL hindi\Train"
test_path = r"C:\Users\user\Downloads\DL hindi\Test"
 # generating test and train data with tensorflow
train_data = tf.keras.preprocessing.image_dataset_from_directory(train_path, image_size=(32, 32), batch_size=32, label_mode='categorical', shuffle=True, interpolation="area")

test_data = tf.keras.preprocessing.image_dataset_from_directory(test_path, image_size=(32, 32), batch_size=32, label_mode='categorical', shuffle=True, interpolation="area")
print(len(test_data.class_names))
class_names = train_data.class_names
print(class_names)

#ARCHITECTURE 1
from tensorflow.keras.layers import Input, Conv2D, Dense, Flatten, Dropout, MaxPool2D, BatchNormalization
from tensorflow.keras.models import Model
from tensorflow.keras.optimizers import SGD
i = Input(shape=(32, 32, 3))
x = Conv2D(64, (3, 3), activation='relu', padding="same")(i)
x = BatchNormalization()(x)
x = Conv2D(64, (3, 3), activation='relu', padding="same")(x)
x = BatchNormalization()(x)
x = MaxPool2D(pool_size=(2, 2), padding='same')(x)
x = Conv2D(128, (3, 3), activation='relu', padding="same")(x)
x = BatchNormalization()(x)
x = Conv2D(128, (3, 3), activation='relu', padding="same")(x)
x = BatchNormalization()(x)
x = MaxPool2D(pool_size=(2, 2), padding='same')(x)
x = Conv2D(256, (3, 3), activation='relu', padding="same")(x)
x = BatchNormalization()(x)
x = MaxPool2D(pool_size=(2, 2), padding='same')(x)
x = Flatten()(x)
x = Dropout(0.4)(x) 
x = Dense(512, activation='relu')(x)
x = BatchNormalization()(x)
x = Dropout(0.4)(x)
x = Dense(36, activation='softmax')(x)
model = Model(i, x)
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
r = model.fit(train_data, validation_data = test_data, epochs = 10)

 # evaluate the model
model.evaluate(train_data)
model.evaluate(test_data)


# ploting the accuracy and loss
import matplotlib.pyplot as plt
train_acc = r.history['accuracy']
val_acc = r.history['val_accuracy']

# plot the accuracy
plt.plot(train_acc, label='Training accuracy')
plt.plot(val_acc, label='Validation accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.show()

# plot the loss
train_loss = r.history['loss']
val_loss = r.history['val_loss']
plt.plot(train_loss, label='Training loss')
plt.plot(val_loss, label='Validation loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.show()


import numpy as np
import cv2
from sklearn.metrics import confusion_matrix
import seaborn as sns
import keras
# get the predicted classes
predicted_probs = model.predict(test_data)
predicted_classes = np.argmax(predicted_probs, axis=1)
# get the true classes
true_classes = np.concatenate([y for x, y in test_data], axis=0)
true_classes = np.argmax(true_classes, axis=1)
# get the class labels
class_labels = list(test_data.class_names)
# get the confusion matrix
cm = confusion_matrix(true_classes, predicted_classes)
# plot the confusion matrix
plt.figure(figsize=(25, 25))
sns.heatmap(cm, annot=True, cmap=plt.cm.Purples, xticklabels=class_labels, yticklabels=class_labels)
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()

#IMAGE PREDICTION
def predict_and_display(image_path, model, class_names):
    image = cv2.imread(image_path)
    resized_image = cv2.resize(image, (32, 32))
    image_array = np.array(resized_image)
    image_array = np.expand_dims(image_array, axis=0)
    prediction = model.predict(image_array)
    predicted_class = class_names[np.argmax(prediction)]
    confidence = np.max(prediction)
    plt.imshow(image[:,:,::-1])
    plt.axis('off')
    plt.title(f'Predicted label: {predicted_class}\nConfidence: {confidence:.2f}')
    plt.show()


image_path = r"C:\Users\user\Desktop\dl hindi\consonants\na.png"
predict_and_display(image_path, model, class_names)
 
```

```python
#ARCHITECTURE 2
from tensorflow.keras.layers import Input, Conv2D, Dense, Flatten, Dropout, MaxPool2D, BatchNormalization
from tensorflow.keras.models import Model
from tensorflow.keras.optimizers import SGD
i = Input(shape = (32, 32, 3))
x = Conv2D(32, (3, 3), strides = 2, activation = 'relu')(i)
x = BatchNormalization()(x)
x = Conv2D(64, (3, 3), strides = 2, activation = 'relu')(x)
x = BatchNormalization()(x)
x = MaxPool2D(pool_size=(2, 2), padding = 'same')(x)
x = Conv2D(128, (3, 3), strides = 2, activation = 'relu')(x)
x = BatchNormalization()(x)
x = MaxPool2D(pool_size=(2, 2), padding = 'same')(x)
x = Flatten()(x)
x = Dropout(0.3)(x)
x = Dense(512, activation = 'relu')(x)
x = BatchNormalization()(x)
x = Dropout(0.3)(x)
x = Dense(36, activation='softmax')(x)
model = Model(i, x)
model.compile(optimizer=SGD(learning_rate=0.001, momentum=0.9),loss='categorical_crossentropy', metrics=['accuracy'])
r = model.fit(train_data, validation_data = test_data, epochs = 10)

model.evaluate(test_data)
model.evaluate(train_data)

# ploting the accuracy and loss
import matplotlib.pyplot as plt
train_acc = r.history['accuracy']
val_acc = r.history['val_accuracy']

# plot the accuracy
plt.plot(train_acc, label='Training accuracy')
plt.plot(val_acc, label='Validation accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.show()

# plot the loss
train_loss = r.history['loss']
val_loss = r.history['val_loss']
plt.plot(train_loss, label='Training loss')
plt.plot(val_loss, label='Validation loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.show()


import numpy as np
import cv2
from sklearn.metrics import confusion_matrix
import seaborn as sns
import keras

# get the predicted classes
predicted_probs = model.predict(test_data)
predicted_classes = np.argmax(predicted_probs, axis=1)

# get the true classes
true_classes = np.concatenate([y for x, y in test_data], axis=0)
true_classes = np.argmax(true_classes, axis=1)

# get the class labels
class_labels = list(test_data.class_names)

# get the confusion matrix
cm = confusion_matrix(true_classes, predicted_classes)

# plot the confusion matrix
plt.figure(figsize=(25, 25))
sns.heatmap(cm, annot=True, cmap=plt.cm.Purples, xticklabels=class_labels, yticklabels=class_labels)
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
import cv2
import numpy as np
import matplotlib.pyplot as plt


#IMAGE PREDICTION
def predict_and_display(image_path, model, class_names):
image = cv2.imread(image_path)
resized_image = cv2.resize(image, (32, 32))
image_array = np.array(resized_image)
image_array = np.expand_dims(image_array, axis=0)
prediction = model.predict(image_array)
predicted_class = class_names[np.argmax(prediction)]
confidence = np.max(prediction)
plt.imshow(image[:,:,::-1])
plt.axis('off')
plt.title(f'Predicted label: {predicted_class}\nConfidence: {confidence:.2f}')
plt.show()


image_path = r"C:\Users\user\Desktop\dl hindi\consonants\na.png"
predict_and_display(image_path, model, class_names)
```
